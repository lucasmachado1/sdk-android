# 📱 AcoNuWiFi — Compilação Automática do APK Android
Este notebook compila o aplicativo nativo Python/KivyMD em um arquivo `.apk` pronto para instalar no seu celular Android.

```python
# 1. Instalar dependências do sistema Linux e Buildozer
!sudo apt-get update -qq
!sudo apt-get install -y python3-pip build-essential git zip unzip openjdk-17-jdk autoconf libtool pkg-config zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev automake ccache lld
!pip install buildozer cython==0.29.36
!mkdir -p /content/aconuwifi
```

```python
# 2. Criar main.py no Colab
import os
from google.colab import files

print('Envie os arquivos main.py e buildozer.spec da pasta do seu computador:')
uploaded = files.upload()
for fn in uploaded.keys():
  os.rename(fn, f'/content/aconuwifi/{fn}')
  print(f'Arquivo {fn} movido para /content/aconuwifi/')
```

```python
# 3. Executar compilação do APK
%cd /content/aconuwifi
!rm -rf .buildozer
!buildozer android debug
```

```python
# 4. Baixar APK para o seu celular/computador
import glob
from google.colab import files

apks = glob.glob('/content/aconuwifi/bin/*.apk')
for apk in apks:
    print(f'Baixando APK: {apk}')
    files.download(apk)
```
